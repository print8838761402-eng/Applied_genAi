# **loop with conditional statement**

```python
# print the values from 1 to 10 without 7
```

```python
# 1,2,3,4,6,7,8,9,10
```

```python
for i in range(1,11,1):
# print(i)
  if i==7:
   continue
  print(i)
```

```python
for i in range(1,11,1):
# print(i)
  if i==5:
   continue
  else:
    print(i)

```

```python
for i in range(1,11,1):
# print(i)
  if i==5:
   break
  else:
    print(i)

```

```python
# print 0,1,3,4
```

```python
for i in range(5):
  if i==2:
    continue
  else:
    print(i)

```
